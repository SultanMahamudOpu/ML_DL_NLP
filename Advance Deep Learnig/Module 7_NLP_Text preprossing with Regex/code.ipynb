# Escape Sequence

````python
txt = 'Hi, This is Sultan, My age is 25 5.'
x = re.findall(r'.',txt) # সব কিছু select করে
print(x)

x = re.findall(r'\.',txt) # যদি শুধু . select করতে চাই
print(x)

x = re.findall(r'\d',txt) # যদি শুধু number আলাদা আলাদা ভাবে select করতে চাই
print(x)

x = re.findall(r'\d+',txt) # যদি শুধু number যেভাবে আছে সেভাবে select করতে চাই
print(x)

x = re.findall(r'\w',txt) # যদি শুধু word select করতে চাই, space, newline বাদে
print(x)

x = re.findall(r'\s',txt) # যদি শুধু space, newline select করতে চাই
print(x)

x = re.findall(r'\D',txt) # যদি number বাদে সব কিছু select করতে চাই
print(x)

x = re.findall(r'\W',txt) # যদি word বাদে সব কিছু select করতে চাই
print(x)

x = re.findall(r'\S',txt) # যদি space বাদে সব কিছু select করতে চাই
print(x)
````

# Digit character class

# Wildcard character .

````python
txt = "hello hlleo hiilo oiiih hwwwo"
x = re.findall(r".",txt) # আমি যদি সব কিছু select করতে তাহলে এইভাবে দিব
print(x)

x = re.findall(r'h...o',txt) # 1st chr "h" আর last char "o" মাঝে 3 টা char যাই থাক select করবে
print(x)

txt = "abc@gmail.com bca@gmail.com aabbbbba@gmail.com aaa@gmail.com"
x = re.findall(r"...@.....\....",txt) # fix gmail select করতে তাহলে এইভাবে দিব। @gmail.com এর আগে বড় থাকলেও সেখান থেকে শুধু ৩ টাই নিবে
print(x)
````

# Asterics

````python
txt = "abc@gmail.com bca@gmail.com aabbbbba@gmail.com aaa@gmail.com aaa@yahoo.com aaa@yahoo.2665656"
x = re.findall(r".*@.*\..*",txt) # variable size gmail select করতে তাহলে এইভাবে দিব। 

txt = "abc@gmail.com bca@gmail.com aabbbbba@gmail.com aaa@gmail.com aaa@yahoo.com aaa@yahoo.2665656"
x = re.findall(r".*@.*.com&*",txt) # শুধু variable size gmail select করতে তাহলে এইভাবে দিব। 
print(x)

txt = "abc@@@gmail.com bca@gmail.com aabbbbba@gmail.com aaa@gmail.com aaa@yahoo.com aaa@yahoo.2665656"
x = re.findall(r".*@.*.com&*",txt) # বেশি @ থাকলেও নিয়ে নিচ্ছে, এটা solve করতে হবে
print(x)
````

````python
txt = "akash@@@gmail.com bca@gmail.com aabbbbba@gmail.com aaa@gmail.com aaa@yahoo.com aaa@yahoo.2665656"

x = re.findall(r'\b[a-zA-Z0-9._%+-]+@[a-zA-Z0-9.-]+\.com\b', txt)

print(x)
````

# Plus Quantifier

````python
txt = "heo"
txt1 = "hello"
x = re.findall(r"he.+o",txt) # * দিলে 0 বা more occurance সব select করে
y = re.findall(r"he.+o",txt1)# + দিলে 1 বা more occurance সব select করে। যদি 1 টা char না থাকে তবে সেটাকে select করবে না। যেমন heo কে করে নাই           
print(x, y)
````

# Question mark

````python
txt = "hello helo"

x = re.findall(r"hel?o",txt) # ? এর আগের 1 বা 0 বার থাকলে print করবে

print(x)
````

````python
txt = "heo hello he1o"

x = re.findall(r"he.?o",txt) # এখানে hello তে ২ টা char আছে তাই সেটা select হয় নাই

print(x)
````

# Capture and group

````python
txt = 'ababab'
x = re.findall(r'ab+', txt)
print(x)
````

````python
txt = 'ababab'
x = re.findall(r'(ab)+', txt) 
print(x)
````

````python
# email বা কোন লিখাকে seperate করা

txt = 'abc@gmail.com'
x = re.findall(r'(\w+)@(\w+)\.com', txt) # tuple return করে
print(x)

for first, second in x:
    print(first, second)
````

````python
txt = 'john 25, Alice 30, Bob 18'
x = re.findall(r'(\w+)\s(\d+)', txt)
print(x)

for name, age in x:
    print(name, age)
````

## Findall

````python
txt = "My lucky numbers are 7,25,99"
x = re.findall(r'\d+', txt)
print(x)
````

# Search

````python
txt = "My lucky number is 25 19 22 1"
x = re.search(r'\d+', txt) # first match টা কে object আকারে ফেরত দেয়। পরের গুলা নেয় না।
                           # আর findall এ সব match গুলাই নেয়
print(x) # পুরা object টা 
print(x.start()) # object এর first টা
print(x.end())   # object এর last টা
print(x.group()) # যেই value টা চাচ্ছি সেটা return করবে। মানে 1st match number টা
````

````python
# email
txt = "contact us at support@phitron.io , you can also email us admin@yahoo.com"

match = re.search(r"\w+@\w+\.\w+",txt)

print(match.group())
````

## Split

````python
txt = "apple, banana, orange"

x = re.split(r",",txt) # কমা এর উপরে ভিত্তি করে split
print(x)

x = re.split(r"\s",txt) # space এর উপরে ভিত্তি করে split
print(x)
````

## Sub

````python
txt = "I have 25 cats and 3 dogs"
x = re.sub(r"\d+","#",txt) # কোন কিছুকে অন্য কিছু দিয়ে replace করে দেয়া

print(x)
````

## Compile

``` 
complie দিয়ে কোন pattern বের করলে সেটা আর বার বার লিখা লাগবে না। যেমন আমাদের বার বার re.findall(r'\d+', txt) লিখা লাগতেছিল। এখানে  r'\d+' এই pattern টা একবার diglar করে দিলে পরে আর লিখতে হবে না 
```

````python
txt = "I have 2 cats and 15 dogs"
````

````python
pattern = re.compile(r"\d+") 
````

````python
#1: compile + findall()
x = pattern.findall(txt)
print(x)
````

````python
#2 compile + search
x = pattern.search(txt)
print(x)
print(x.start())
print(x.end())
print(x.group())
````

````python
#Compile + sub
x = pattern.sub('#', txt)
print(x)
````

````python
# Compile + split

pattern = re.compile(r",")

txt = "apple, banana, orange"

x = pattern.split(txt)

print(x)
````
