## Remove HTML Tags

````python
import re

def remove_html_tags(text): # আমরা regular experation এর জন্য function দিয়ে কাজ করব, তাহলে সুবিধা হবে
    pattern = re.compile(r'<.*?>')
    return pattern.sub(r'', text)
````

## Remove punctuations


````python
import string
````

````python
exclude = string.punctuation
````

````python
exclude
````

### way 1, slow

````python
def remove_punc1(text):
    for char in exclude:
        text = text.replace(char, '')
    return text
````

````python
remove_punc1('Hi, I am Sultan.Mahamud.Opu?!')
````

### way 2, fast

````python
def remove_punc(text):
    return text.translate(str.maketrans(' ', ' ', exclude))
````

````python
remove_punc('Hi, I am Sultan.Mahamud.Opu?!')
````

````python
df['review'] = df['review'].apply(remove_punc)
````

# Stopwords removal

````python
text = """
Yesterday, I was working on a Machine Learning project in PHITRON, and I was trying to understand how the model is performing on the dataset. The dataset is quite large, and it contains a lot of reviews that are written by users from different countries.

I noticed that the model is not performing well for some of the samples, especially those that are very short or have too many stopwords in them. Because of this, I decided to clean the text data and remove unnecessary words that do not contribute much to the meaning of the sentence.

While I was doing this, I also checked how the preprocessing steps like tokenization, lowercasing, and punctuation removal are affecting the dataset. It was interesting to see that after removing stopwords, the text becomes more meaningful and easier for the model to understand.

In addition to that, I was also reading some documentation from https://scikit-learn.org/stable/ and watching tutorials on https://www.youtube.com while comparing different NLP techniques. The goal is to improve the accuracy of the model and make it more robust for real-world applications.

Overall, I think that removing stopwords is an important step in the NLP pipeline because it helps reduce noise and improve performance of machine learning models.
"""
````

````python
import nltk
from nltk.corpus import stopwords
````

````python
nltk.download('stopwords')
````

````python
stop_words = set(stopwords.words('bengali')) # fast access এর জন্য set use করছি
````

````python
stop_words
````

````python
stop_words = set(stopwords.words('english')) 
````

````python
stop_words
````

````python
def remove_stopwords(text):
    words = text.split()
    cleaned = []
    
    for word in words:
        word = word.lower().strip(string.punctuation)
        
        if word and word not in stop_words:
            cleaned.append(word)
    
    return ' '.join(cleaned)
````

````python
remove_stopwords(text)
````

````python
df['review'] = df['review'].apply(remove_stopwords)
df['review']
````

# Emoji removal

````python
text_with_emoji = '''Yesterday I completed my Machine Learning project on sentiment analysis 😊🔥.
The model performed really well on the IMDB dataset 👍🎯.
I learned a lot about preprocessing steps like tokenization, stopwords removal, and stemming 📚💡.
However, debugging errors was quite frustrating 😓💻.
Overall, the experience was amazing and I feel more confident now 🚀😄.'''
````

````python
import re

def remove_emoji(text):
    emoji_pattern = re.compile(
        "["
        "\U0001F600-\U0001F64F"  # emoticons
        "\U0001F300-\U0001F5FF"  # symbols & pictographs
        "\U0001F680-\U0001F6FF"  # transport & map symbols
        "\U0001F1E0-\U0001F1FF"  # flags
        "\U00002700-\U000027BF"  # dingbats
        "\U000024C2-\U0001F251"
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub(r'', text)
````

````python
clean_text = remove_emoji(text_with_emoji)
print(clean_text)
````

````python
df['review'] = df['review'].apply(remove_emoji)
df['review']
````

# Stemming

````python
from nltk.stem.porter import PorterStemmer
````

````python
ps = PorterStemmer()

def stem_words(text):
    return ' '.join([ps.stem(word) for word in text.split()])
````

````python
sample = "walk walks walking walked"
stem_words(sample)
````

# Lemmatization

```
stemming এমনি কোন word ছোট করে দেয়। সেটা dictionary word কি না, সেটা দেখে না।
lemmatization কোন word এর correct dictionary base form এ নিয়ে আসে।
lematization অবশ্যই ভালো, তবে এটায় time বেশি লাগে
```

````python
import  nltk
from nltk.stem import WordNetLemmatizer
````

````python
nltk.download('wordnet')
````

````python
wordnet_nemmatizer = WordNetLemmatizer()
````

````python
sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
````

````python
sentence = remove_punc(sentence)
````

````python
sentence = sentence.split()
sentence
````

````python
print("{0:20}{1:20}".format("Word","Lemma")) # {0:20}{1:20} এগুলা টেবিলের মত format দেখানোর জন্য আর কিছুই না

for word in sentence:
    print('{0:20}{1:20}'.format(word, wordnet_nemmatizer.lemmatize(word, pos='v')))
````
