## Q1: One-Hot Encoding

## Q2: TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love NLP",
    "NLP is amazing",
    "Machine learning is fun"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print(tfidf_matrix.toarray())
```

## Q11: Stopword Removal

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')

words = ["I", "love", "to", "learn", "NLP"]

stop_words = set(stopwords.words("english"))

filtered = [word for word in words if word.lower() not in stop_words]

print(filtered)
```

## Q12: Stemming

## Q13: Lemmatization

# Word2Vec: CBOW vs Skip-Gram

**1. How does each model predict words?**

* **CBOW (Continuous Bag of Words):** It predicts the **target word from its surrounding context words**.

  * Example: `I ___ to school` → predicts **go** using the surrounding words.

* **Skip-Gram:** It predicts the **surrounding context words from a target word**.

  * Example: `I go to school` → using **go**, it predicts words such as **I, to, school**.

**2. Which model performs better on small datasets?**

**CBOW** generally performs better on **small datasets** because it is faster to train and learns effectively from the available context.

**3. Which model is better for learning rare words?**

**Skip-Gram** is better for learning **rare words** because it trains on individual target-context word pairs and can learn useful representations even when a word appears infrequently.

**4. One practical advantage of Word2Vec over One-Hot Encoding**

Word2Vec represents words as **dense, low-dimensional vectors** that can capture **semantic similarity and relationships between words**. In contrast, One-Hot Encoding produces sparse vectors and does not capture relationships between words.

**Summary:**

| Feature        | CBOW               | Skip-Gram                       |
| -------------- | ------------------ | ------------------------------- |
| Prediction     | Context → Target   | Target → Context                |
| Small datasets | Better             | Less effective                  |
| Rare words     | Less effective     | Better                          |
| Training       | Faster             | Slower                          |
| Main advantage | Efficient learning | Better rare-word representation |
