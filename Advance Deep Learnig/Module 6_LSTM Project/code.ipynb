```python
def text_to_indices(sentence, vocab):
    numerical_sentence= []
    for token in sentence:
        if token in vocab:
            numerical_sentence.append(vocab[token])
        else:
            numerical_sentence.append(vocab['<unk>'])
            
    return numerical_sentence
```

```python
input_numerical_sentence= []
for sentence in input_sentences:
    input_numerical_sentence.append(text_to_indices(word_tokenize(sentence.lower()), vocab))
```

# Custom Dataset and Dataloader

```python
class CustomDataset(Dataset):
    def __init__(self, X, y):
        self.X= X
        self.y= y
        
    def __len__(self):
        return self.X.shape[0]
    
    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]
        
```

```python
dataset= CustomDataset(X, y)
```

```python
len(dataset)
```

```python
dataloader= DataLoader(dataset, batch_size=32, shuffle=True)
```

# LSTM Cell

```python
class LSTMModel(nn.Module):
    def __init__(self, vocab_size):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, 100)
        self.lstm = nn.LSTM(100, 150, batch_first=True) # batch_first=true করে দিলে batch_size, sequence_len, input_size হয়ে যায়। যার ফলে shape miss mass এর problem টা solve হয়ে যায়
        self.fc = nn.Linear(150, vocab_size
                            )
    def forward(self, x):
        embedded = self.embedding(x)
        intermediate_hidden_states, (final_hidden_state, final_cell_state) = self.lstm(embedded)
        output = self.fc(final_hidden_state.squeeze(0))
        return output
```

```python
model = LSTMModel(len(vocab))
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
```

```python
model.to(device)
```

# Training Loop

```python
epochs = 50
learning_rate = 0.001
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

for epoch in range(epochs):
    total_loss = 0
    
    for batch_x, batch_y in dataloader:
        batch_x, batch_y = batch_x.to(device), batch_y.to(device)
        optimizer.zero_grad()
        output = model(batch_x)
        loss = criterion(output, batch_y)
        loss.backward()
        optimizer.step()
        total_loss = total_loss + loss.item()
        
    print(f'Epoch: {epoch + 1}, Loss: {total_loss: .4f}')
```

# Prediction

```python
def prediction(model, vocab, text):
    # tokenize
    tokenized_text = word_tokenize(text.lower())
    
    # numerical indices
    numerical_text = text_to_indices(tokenized_text,vocab)
    
    # padding
    padded_text = torch.tensor([0]* (25-len(numerical_text)) + numerical_text , dtype=torch.long).unsqueeze(0)
    
    # টেনসরটিকে মডেলে পাঠানোর আগে device-এ (GPU) ট্রান্সফার করুন
    padded_text = padded_text.to(device)
    
    # send to model
    output = model(padded_text)
    
    # predicted index
    value, index = torch.max(output,dim=1)
    
    # merge with text (GPU tensor থেকে পাইথন ইন্টিজারে নিতে .item() ব্যবহার করা হয়েছে)
    return text + " " + list(vocab.keys())[index.item()]
```

# Accuracy Calculation

```python
def calculate_accuracy(model, dataloader, device):
    model.eval()
    correct = 0
    total = 0
    
    with torch.no_grad():
        for batch_x, batch_y in dataloader:
            batch_x, batch_y = batch_x.to(device), batch_y.to(device)
            
            # get moel prediction
            output = model(batch_x)
            
            # et the predictec word indices
            _, predicted = torch.max(output, dim=1)
            
            # compare with actual label
            correct += (predicted == batch_y).sum().item()
            total += batch_y.size(0)
            
        accuracy = correct / total * 100
        return accuracy
```

```python
accuracy = calculate_accuracy(model, dataloader, device)
print(f'Model accuracy {accuracy: .4f}%')
```
