TF -> একটা word ওই document এ কয়বার আছে, যে যত বেশি আছে তার value বেশি

IDF -> একটা word ওই documnet এ কতটা rare, মানে যে যত কম বার আসছে তার value তত বেশি

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
corpus = [
    "students watch youtube",
    "youtube watch youtube",
    "students write comments",
    "youtube write comments"
]
```

```python
df = pd.DataFrame({
    'text' : corpus,
    'output' : [1, 1, 0, 0]
})

print(df)
```

```python
tfidf = TfidfVectorizer()
```

```python
X = tfidf.fit_transform(df['text'])
```

```python
# convert sparse matrix to dataframe

tfidf_df = pd.DataFrame(
    X.toarray(),
    columns= tfidf.get_feature_names_out()
)

print(tfidf_df)
```

```python
print("\nVocabulary:")
print(tfidf.vocabulary_)
```

```python
print("\nFeature Names:")
print(tfidf.get_feature_names_out())
```

```python
print("\nIDF Values:")
for word, idf in zip(tfidf.get_feature_names_out(), tfidf.idf_):
    print(f"{word:10} -> {idf:.4f}")
```

```python
new_text = ["people watch comment"]

new_vector = tfidf.transform(new_text)

print("\nNew Sentence:")
print(new_text[0])

print("\nTF-IDF Vector for New Sentence:")
print(pd.DataFrame(
    new_vector.toarray(),
    columns=tfidf.get_feature_names_out()
))
```

```python
#  Example with OOV Word
new_text_oov = ["people watch youtube"]

new_vector_oov = tfidf.transform(new_text_oov)

print("\nSentence with New Word:")
print(new_text_oov[0])

print("\nTF-IDF Vector:")

print(pd.DataFrame(
    new_vector_oov.toarray(),
    columns=tfidf.get_feature_names_out()
))
```
