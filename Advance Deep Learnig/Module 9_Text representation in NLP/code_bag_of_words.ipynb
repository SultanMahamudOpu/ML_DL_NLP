````python
from sklearn.feature_extraction.text import  CountVectorizer
import pandas as pd
````

````python
corpus = [
    "students watch youtube youtube youtube",
    "youtube watch youtube",
    "students write comments",
    "youtube write comments"
]
````

# create bag of words model

````python
cv = CountVectorizer()
````

# learn vocabulary and transfrom documnent


````python
X = cv.fit_transform(corpus)
````

#show vocabulary

````python
cv.get_feature_names_out()
````

# show bag of words matrix

````python
X.toarray()
````

````python
bow_df = pd.DataFrame(
    X.toarray(),
    columns = cv.get_feature_names_out(),
    index = ['D1', 'D2', 'D3', 'D4']
)

bow_df
````

```
এখানে কোন word কয়বার আছে সেই frequency টা দেখাচ্ছে। আমরা যদি চাই ওই word টা শুধু আছে কি নাই, সেটা দেখতে তাহলে শুধু "binary=True" করে দিব
```

````python
cv = CountVectorizer(binary = True)
````

````python
X = cv.fit_transform(corpus)
````

````python
bow_df = pd.DataFrame(
    X.toarray(),
    columns = cv.get_feature_names_out(),
    index = ['D1', 'D2', 'D3', 'D4']
)

bow_df
````

# n_gram

````python
cv = CountVectorizer(ngram_range=(2, 2)) # ngram_range=(min, max) এত গুলা word এক সাথে নিয়ে bag বানাইতে চাই সেটা। যদি ngram_range=(1, 1) দেই তবে সেটা হল bag of word
````

````python
X = cv.fit_transform(corpus)
````

````python
cv.get_feature_names_out()
````

````python
bow_df = pd.DataFrame(
    X.toarray(),
    columns = cv.get_feature_names_out(),
    index = ["D1","D2","D3","D4"]
)
bow_df
````

````python
cv = CountVectorizer(ngram_range=(1, 3))
````

````python
X = cv.fit_transform(corpus)
````

````python
cv.get_feature_names_out()
````

````python
bow_df = pd.DataFrame(
    X.toarray(),
    columns = cv.get_feature_names_out(),
    index = ["D1","D2","D3","D4"]
)
bow_df
````
